Going to learn about merging data 
4 possible ways to merge data: 

(1) Inner join (INNER JOIN): Joins data when both have keys.

(2) Left outer join (LEFT JOIN): Joins when the data on the left side has a key.

(3) Right outer join (RIGHT JOIN): Joins when the key of the data is on the right side.

(4) Full outer join (FULL JOIN): Joins when the key exists on either side.



```python
# lets merge using inner join on the MAL_ID 
pd.merge(anime_data_extraced,anime_synop, on = "MAL_ID")
```

Mapping

```python
# Reference data
status_map = {
    1: "Watching",
    2: "Completed",
    3: "On-Hold",
    4: "Dropped",
    6: "Plan to Watch"
}
status_map
```

```python
# Drop invalid rows first
invalid_status = [0, 5, 33, 55]
# Use ~ to represent NOT
anime_list = anime_list[~anime_list['watching_status'].isin(invalid_status)]
anime_list['status_label'] = anime_list['watching_status'].map(status_map)
anime_list
```

```python
# We can also use .map() to apply certain fuctions 
anime_data_extraced["Score"] = anime_data_extraced["Score"].astype(float)

anime_data_extraced['ScoreCategory'] = anime_data_extraced['Score'].map(
    lambda x: 'Low' if x < 5 else ('Average' if x <= 7 else 'High')
)
anime_data_extraced.head()
```

```python
# To make the continous data to discrete data we use .cut()
scores = [0, 2, 4, 6, 8, 10]
anime_data_cut_by_score = pd.cut(anime_data_extraced["Score"], scores)
anime_data_cut_by_score
```

```python
pd.value_counts(anime_data_cut_by_score)
```

```python
# lets add labels
group_names = ['D', 'C', 'B', 'A', 'A+']
anime_data_cut_by_score = pd.cut(anime_data_extraced["Score"], scores, labels=group_names)
pd.value_counts(anime_data_cut_by_score)
```

```python
# Time to aggregate data 
anime_data_extraced.groupby('Type').size()
```

```python
# calculate the mean of the score for every type
anime_data_extraced.groupby('Type')['Score'].mean()
```

```python
# the groupby is an iterator
for group, subdf in anime_data_extraced.groupby('Type'):
    print('==========================================================')
    print('Type:{0}'.format(group))
    print(subdf)
```
